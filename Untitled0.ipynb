4- Criar uma matriz com 50 elementos e 03 dimensões.

5- Criar um gráfico pizza com 03 itens destacando um item

```python
# Definindo variáveis
df02= pd.DataFrame ({'Fatias': [8,12,10], 'Atividades': ['Lazer', 'Estudo', 'Viagem']}, index= ['r', 'm', 'y'])
df02

```

```python

# Criando um gráfico
fatias = [8, 12, 10]
atividades = ['Lazer', 'Estudo', 'Viagem']
colunas = ['r', 'm', 'y']

plt.figure()
plt.pie(fatias, labels = atividades, colors = colunas, startangle = 90, shadow = True,
explode = (0, 0.1, 0))
plt.show()




```

6- Criar um gráfico tipo linha


```python

# Definindo variáveis
x = [1, 3, 5]
y = [1, 2, 5]


# Criando um gráfico
plt.plot(x, y)


```

```python




# Atribuindo um título e legenda ao gráfico
plt.title('Gráfico de linhas')
plt.xlabel('Variável 01')
plt.ylabel('Variável 02')

plt.plot(x, y, label= 'Aprendendo gráfico de linhas, novembro de 2025')
plt.legend()

plt.show()
```

7- Criar um gráfico com pontos em formato de estrela

```python
# Definindo variáveis
x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [5, 2, 4, 5, 6, 8, 4, 8]

# Criando um gráfico
plt.scatter(x, y, label = 'Pontos', color = 'b', marker = '*', s = 100)
plt.legend()
plt.show()
```
